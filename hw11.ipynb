## 1 Functions
**Orbital Periods and Kepler's Third Law** \
Kepler's Third Law states that the square of the orbital period $T$ (in years) of a planet is proportional to the cube of the semi-major axis $a$ (in astronomical units, AU):

$$
T^2 = a^3
$$

Write a **function** that calculates the orbital period $T$ given the semi-major axis $a$. (So, you can assume that $T = \sqrt{a^3}$.)

```python
# Your code here (WITHOUT NUMPY! or any other modules/libraries/packages)

def orbital_period(a):
    T = (a**3)**(1/2)
    return T

print(orbital_period(2))
```

Call your function with the following values: 
* `planet_a` $ = 0.5\,\text{AU}$
* `planet_b` $ = 1.0\,\text{AU}$
* `planet_c` $ = 1.5\,\text{AU}$

```python
# Your code here
planet_a = 0.5
planet_b = 1.0
planet_c = 1.5

print("Planet A orbital period is:", orbital_period(planet_a))
print("Planet B orbital period is:", orbital_period(planet_b))
print("Planet C orbital period is:", orbital_period(planet_c))
```

## 4 Plotting
Create a scatter plot of the following semi-major axes (x-axis) vs. orbital periods (y-axis). Label the axes, add a legend and add a title: "Orbital Periods of Exoplanets."

`semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]`

```python
# Your code here
import matplotlib.pyplot as plt

semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]
orbital_periods = [a**(3/2) for a in semi_major_axes]

plt.scatter(semi_major_axes, orbital_periods, label="Exoplanets", color="blue")

plt.xlabel("Semi-major Axes")
plt.ylabel("Orbital Periods (years)")
plt.title("Orbital Periods of Exoplanets")
plt.legend()

plt.grid(True)
plt.show()
```

Create a bar plot of planet radii vs. their equilibrium temperatures. Label the axes, add a legend and add a title: "Equilibrium Temperatures vs. Radii."

```python
# Your code here

plt.bar(radii, temperatures, label="Equilibrium Temperatures", color="Blue")

plt.xlabel("Planet Radii")
plt.ylabel("Equilibrium Temperatures")
plt.title("Equilibrium Temperatures vs. Radii.")
plt.legend()

plt.grid(axis='y', linestyle='--', alpha=0.7)  #  grid lines for some clarity
plt.show()
```

Put the scatter plot and bar plot on the same figure (horizonatally). 

```python
# Your code here
#subplots and stuff

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].scatter(semi_major_axes, orbital_periods, color='blue', label="Exoplanets")
axes[0].set_xlabel("Semi-Major Axes")
axes[0].set_ylabel("Orbital Periods (years)")
axes[0].set_title("Orbital Periods of Exoplanets")
axes[0].legend()
axes[0].grid(True)

axes[1].bar(radii, temperatures, color='orange', label="Equilibrium Temperatures")
axes[1].set_xlabel("Planet Radii (Earth Radii)")
axes[1].set_ylabel("Equilibrium Temperatures")
axes[1].set_title("Equilibrium Temperatures vs. Radii")
axes[1].legend()
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
