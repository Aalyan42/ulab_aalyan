# 1) Recap of lecture
**Copy and paste** your code from lecture for the neural network with the **default** settings (i.e. before we improved the model). Make sure to also import the `data.py` file. 

```python
import data
import matplotlib.pyplot as plt
import torch 
```

```python

model = torch.nn.Sequential(
    torch.nn.Linear(2, 50), # Input layer -> Hidden layer
    torch.nn.Sigmoid(), # Activation function
    torch.nn.Linear(50, 1) # Hidden layer -> Output layer
)
```

```python

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for t in range(1000):
    y_pred = model(data.x)            # Make a prediction
    loss = loss_fn(y_pred, data.y)    # Compare prediction to actual value
    optimizer.zero_grad()             # Reset gradients
    loss.backward()                   # Compute gradients
    optimizer.step()                   # Update model

    if t % 100 == 0:
        print(f"Iteration {t}, Loss: {loss.item()}")
```

What is the difference between artificial intelligence, machine learning and deep learning?

```python
# Artificial intelligence is the general umbrella term
# ML is a where algos learn from data to make some decisions, more data even better.
# DL is subset of ML, more complex, and uses neural networks. 
```

What is a neural network?

```python
# a bunch of layer, pass data through it, model goes through layers learns and gives answer
# neurons map inputs to ouputs
```

In lecture, we used `torch.nn.Linear(2, 50)`. What do the numbers `2` and `50` represent? 

```python
# 2 and 50 represent our input and output #s, 2 is # of input features, 50 is # of output features
```

What does the activation function do?

```python
# An activation function introduces non-linearity in the nn, learns complex patterns in the data.
# w/out activation  network would behave like a simple linear model regardless of # of layers.
```

What is a loss function? Why do we need one?

```python
# Loss func is how wrong is the model. We need it so that we can learn and improve on the previous output
```

Why do we call `optimizer.zero_grad()` before calling `loss.backward()`?

```python
# we call the optimizer before the loss because we need to tell the numbers to reset so they don't keep compounding on themselves. Pause after every iteration and try again. 
# then we compute our gradients
```

```python
# simple model

import torch
import torch.nn as nn
import torch.optim as optim
import seaborn as sns
import pandas as pd
```

```python
#  dataset
df = sns.load_dataset("titanic").dropna()  # Remove missing values

X = df[['pclass', 'age', 'fare']]  # numerical stuff
y = df['survived']  # binary 0 = No, 1 = Yes

# tensors
X_tensor = torch.tensor(X.values, dtype=torch.float32)
y_tensor = torch.tensor(y.values, dtype=torch.float32).unsqueeze(1)

```

```python

model = nn.Sequential(
    nn.Linear(3, 5), 
    nn.ReLU(),
    nn.Linear(5, 1), 
    nn.Sigmoid()  
)

# loss func and optimizer
loss_fn = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# train
for t in range(500): 
    y_pred = model(X_tensor) 
    loss = loss_fn(y_pred, y_tensor) 
    optimizer.zero_grad()  
    loss.backward() 
    optimizer.step()  
    
    if t % 100 == 0:
        print(f"Iteration {t}, Loss: {loss.item()}")
```

```python
# got the data from titanic dataset in seaborn.

# used ticket class, age, and fare as input features. The target variable is survived where 1 = survived, 0 = did not survive.

# model matches data pretty well, the loss started very high at 9.16 but quickly dropped to around 0.58 after 400 iterations
#  model is learning as the loss is decreasing over time, but lots of room to improve, only somewhat accurate predictions
```
